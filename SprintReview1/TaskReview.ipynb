```
using System;
using System.Collections.Generic;

// ===== 1. Товар =====
public class Product
{
    // Поля
    private string _name;
    private decimal _price;
    private int _quantity;
    protected string _category;
    internal string _description;

    // Свойства (полные, через приватные поля)
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                Console.WriteLine("Цена не может быть отрицательной!");
        }
    }

    public int Quantity
    {
        get { return _quantity; }
        set
        {
            if (value >= 0)
                _quantity = value;
            else
                Console.WriteLine("Количество не может быть отрицательным!");
        }
    }

    public string Category
    {
        get { return _category; }
        set { _category = value; }
    }

    public string Description
    {
        get { return _description; }
        set { _description = value; }
    }

    // Конструктор
    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    // Методы
    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        int newQuantity = Quantity + amount;
        if (newQuantity >= 0)
            Quantity = newQuantity;
        else
            Console.WriteLine($"Ошибка: количество не может стать отрицательным (текущее: {Quantity}, попытка: {amount}).");
    }

    public string GetProductInfo()
    {
        return $"Товар: {Name} | Категория: {Category} | Цена: {Price:C} | Кол-во: {Quantity} | Описание: {Description} | Сумма: {GetTotalPrice():C}";
    }
}

// ===== 2. Продавец =====
public class Seller
{
    // Поля
    private string _name;
    private string _employeeId;
    protected decimal _salary;
    internal string _contactInfo;
    private List<Product> _products = new List<Product>();

    // Свойства
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string EmployeeId
    {
        get { return _employeeId; }
        set { _employeeId = value; }
    }

    public decimal Salary
    {
        get { return _salary; }
        set
        {
            if (value >= 0)
                _salary = value;
            else
                Console.WriteLine("Зарплата не может быть отрицательной!");
        }
    }

    public string ContactInfo
    {
        get { return _contactInfo; }
        set { _contactInfo = value; }
    }

    // Конструктор
    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    // Методы
    public void AddProduct(Product product)
    {
        _products.Add(product);
        Console.WriteLine($"Продавец {Name} добавил товар «{product.Name}» в свой список.");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (!_products.Contains(product))
        {
            Console.WriteLine($"Продавец {Name} не торгует товаром «{product.Name}».");
            return;
        }

        if (quantity > product.Quantity)
        {
            Console.WriteLine($"Недостаточно товара «{product.Name}» на складе (есть {product.Quantity}, запрошено {quantity}).");
            return;
        }

        product.UpdateQuantity(-quantity);
        Console.WriteLine($"Продавец {Name} продал {quantity} ед. товара «{product.Name}». Остаток: {product.Quantity}.");

        // Если товар закончился — убираем из личного списка продавца
        if (product.Quantity == 0)
        {
            _products.Remove(product);
            Console.WriteLine($"Товар «{product.Name}» распродан и снят с продажи у {Name}.");
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {Name} | ID: {EmployeeId} | Зарплата: {Salary:C} | Контакты: {ContactInfo} | Товаров на продаже: {_products.Count}";
    }

    public List<Product> GetProducts()
    {
        return _products;
    }
}

// ===== 3. Магазин =====
public class Store
{
    // Поля
    private string _storeName;
    private string _location;
    private string _storeHours;
    private List<Product> _products = new List<Product>();
    private List<Seller> _sellers = new List<Seller>();

    // Свойства
    public string StoreName
    {
        get { return _storeName; }
        set { _storeName = value; }
    }

    public string Location
    {
        get { return _location; }
        set { _location = value; }
    }

    public string StoreHours
    {
        get { return _storeHours; }
        set { _storeHours = value; }
    }

    // Конструктор
    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    // Методы
    public void AddSeller(Seller seller)
    {
        _sellers.Add(seller);
        Console.WriteLine($"Магазин «{StoreName}» принял на работу продавца {seller.Name}.");
    }

    public void AddProduct(Product product)
    {
        _products.Add(product);
        Console.WriteLine($"Магазин «{StoreName}» завёл на складе товар «{product.Name}».");
    }

    public void ListProducts()
    {
        Console.WriteLine($"\n=== Товары на складе магазина «{StoreName}» ===");
        if (_products.Count == 0)
        {
            Console.WriteLine("  Склад пуст.");
            return;
        }

        foreach (var product in _products)
        {
            Console.WriteLine("  " + product.GetProductInfo());
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {StoreName} | Адрес: {Location} | Часы работы: {StoreHours} | Продавцов: {_sellers.Count} | Товаров на складе: {_products.Count}";
    }
}

// ===== top-level код (точка входа) =====

// 5 товаров
Product beer = new Product("Пиво Amsterdam Navigator", 89.90m, 100, "Алкоголь", "Светлое крепкое пиво");
Product milk = new Product("Молоко Золотые Луга", 79.50m, 50, "Молочное", "Пастеризованное 3.2%");
Product bread = new Product("Хлеб Дарницкий", 45.00m, 30, "Выпечка", "Нарезной");
Product cheese = new Product("Сыр БрестЛитовск", 320.00m, 15, "Молочное", "Твёрдый, Сливочный");
Product tea = new Product("Чай Richard", 250.00m, 20, "Напитки", "Чёрный листовой");

// 2 продавца
Seller nikitin = new Seller("Никитин Р.Ю.", "EMP-001", 60000m, "+7-900-111-22-33");
Seller petrova = new Seller("Петрова А.С.", "EMP-002", 65000m, "+7-900-444-55-66");

// 1 магазин
Store store = new Store("Красное&Белое", "ул. Луначарского, 47/1", "09:00 — 22:05");

Console.WriteLine("=== Информация о магазине и продавцах ===");
Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(nikitin.GetSellerInfo());
Console.WriteLine(petrova.GetSellerInfo());

Console.WriteLine("\n=== Наполнение склада магазина ===");
store.AddProduct(beer);
store.AddProduct(milk);
store.AddProduct(bread);
store.AddProduct(cheese);
store.AddProduct(tea);

Console.WriteLine("\n=== Продавцы получают товар на продажу ===");
nikitin.AddProduct(beer);
nikitin.AddProduct(milk);
nikitin.AddProduct(bread);

petrova.AddProduct(cheese);
petrova.AddProduct(tea);

Console.WriteLine("\n=== Приём продавцов в магазин ===");
store.AddSeller(nikitin);
store.AddSeller(petrova);

Console.WriteLine("\n=== Обновление информации о магазине ===");
Console.WriteLine(store.GetStoreInfo());

store.ListProducts();

Console.WriteLine("\n=== Продажа: Никитин продаёт 30 бутылок пива Amsterdam Navigator ===");
nikitin.SellProduct(beer, 30);
store.ListProducts();

Console.WriteLine("\n=== Продажа: Петрова распродаёт весь сыр (15 шт.) ===");
petrova.SellProduct(cheese, 15);
store.ListProducts();

Console.WriteLine("\n=== Обновление информации о магазине после продаж ===");
Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(nikitin.GetSellerInfo());
Console.WriteLine(petrova.GetSellerInfo());
```
