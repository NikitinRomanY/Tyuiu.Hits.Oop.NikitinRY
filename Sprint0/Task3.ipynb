Описание задачи:
Создать базовый класс PaymentMethod в C#, который будет представлять
различные способы оплаты. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу PaymentMethod:
• Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты
(MethodName), Минимальная сумма (MinAmount).
• Методы:
o ProcessPayment(decimal amount): метод для обработки платежа
указанной суммы.
o CheckMinimumAmount(decimal amount): метод для проверки
минимальной суммы платежа.
o GetPaymentDetails(): метод для получения деталей способа оплаты.
Требования к производным классам:
1. ОнлайнОплата (OnlinePayment): Должен содержать дополнительные
атрибуты, такие как URL платежной системы (PaymentUrl).
Метод ProcessPayment() должен быть переопределен для включения URL
платежной системы в процесс оплаты.
2. БанковскийПеревод (BankTransfer): Должен содержать дополнительные
атрибуты, такие как Банковские данные (BankData).
Метод CheckMinimumAmount() должен быть переопределен для проверки
минимальной суммы платежа с учетом банковских комиссий.
3. Наличные (CashPayment) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Место выдачи наличных
(CashPickupPoint). Метод GetPaymentDetails() должен быть переопределен
для отображения места выдачи наличных.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый класс — способ оплаты
public class PaymentMethod
{
    // Атрибуты
    public int PaymentMethodId { get; set; }
    public string MethodName { get; set; }
    public decimal MinAmount { get; set; }

    public PaymentMethod(int _paymentMethodId, string _methodName, decimal _minAmount)
    {
        PaymentMethodId = _paymentMethodId;
        MethodName = _methodName;
        MinAmount = _minAmount;
    }

    // Методы
    public virtual void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Базовый] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Базовый] Платёж на сумму {amount} успешно обработан через «{MethodName}».");
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"[Базовый] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}");
    }
}

// Производный класс
public class OnlinePayment : PaymentMethod
{
    // Атрибуты
    public string PaymentUrl { get; set; }

    public OnlinePayment(int _id, string _name, decimal _minAmount, string _paymentUrl)
        : base(_id, _name, _minAmount)
    {
        PaymentUrl = _paymentUrl;
    }

    // Методы
    public override void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Онлайн] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Онлайн] Перенаправление на {PaymentUrl}...");
        Console.WriteLine($"[Онлайн] Платёж на сумму {amount} успешно проведён.");
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Онлайн] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, URL: {PaymentUrl}");
    }
}

// Производный класс
public class BankTransfer : PaymentMethod
{
    // Атрибуты
    public string BankData { get; set; }
    private decimal CommissionRate = 0.02m;

    public BankTransfer(int _id, string _name, decimal _minAmount, string _bankData)
        : base(_id, _name, _minAmount)
    {
        BankData = _bankData;
    }

    // Методы
    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * CommissionRate;
        Console.WriteLine($"[Перевод] Комиссия банка: {commission}. Итого к списанию: {amount + commission}");
        return amount >= MinAmount;
    }

    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"[Перевод] Банковские данные: {BankData}");
        base.ProcessPayment(amount);
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Перевод] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, Реквизиты: {BankData}, Комиссия: {CommissionRate:P0}");
    }
}

// Производный класс
public class CashPayment : PaymentMethod
{
    // Атрибуты
    public string CashPickupPoint { get; set; }

    public CashPayment(int _id, string _name, decimal _minAmount, string _cashPickupPoint)
        : base(_id, _name, _minAmount)
    {
        CashPickupPoint = _cashPickupPoint;
    }

    // Методы
    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Наличные] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, Место выдачи: {CashPickupPoint}");
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Наличные] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Наличные] Получите наличные в пункте: {CashPickupPoint}");
        Console.WriteLine($"[Наличные] Платёж на сумму {amount} подтверждён.");
    }
}

// ==== top-level код (точка входа) ====
PaymentMethod[] methods = new PaymentMethod[]
{
    new OnlinePayment(1, "Карта онлайн", 100m, "https://pay.example.com"),
    new BankTransfer(2, "Банковский перевод", 500m, "IBAN: DE12 3456 7890"),
    new CashPayment(3, "Наличные в кассе", 50m, "Касса №5, ул. Ленина 10")
};

foreach (var method in methods)
{
    Console.WriteLine($"\n══════════ Обработка способа оплаты ID: {method.PaymentMethodId} ══════════");

    Console.WriteLine("1. Детали способа оплаты:");
    method.GetPaymentDetails();

    Console.WriteLine("2. Успешный платёж на 1000:");
    method.ProcessPayment(1000m);

    Console.WriteLine("3. Слишком маленькая сумма (10):");
    method.ProcessPayment(10m);
}

Console.WriteLine("\n═══════════════════════════════════════════════════════════");
```
