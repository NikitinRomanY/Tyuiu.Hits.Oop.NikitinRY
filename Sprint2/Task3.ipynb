Описание задачи:
Создать базовый класс PaymentMethod в C#, который будет представлять
различные способы оплаты. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу PaymentMethod:
• Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты
(MethodName), Минимальная сумма (MinAmount).
• Методы:
o ProcessPayment(decimal amount): метод для обработки платежа
указанной суммы.
o CheckMinimumAmount(decimal amount): метод для проверки
минимальной суммы платежа.
o GetPaymentDetails(): метод для получения деталей способа оплаты.
Требования к производным классам:
1. ОнлайнОплата (OnlinePayment): Должен содержать дополнительные
атрибуты, такие как URL платежной системы (PaymentUrl).
Метод ProcessPayment() должен быть переопределен для включения URL
платежной системы в процесс оплаты.
2. БанковскийПеревод (BankTransfer): Должен содержать дополнительные
атрибуты, такие как Банковские данные (BankData).
Метод CheckMinimumAmount() должен быть переопределен для проверки
минимальной суммы платежа с учетом банковских комиссий.
3. Наличные (CashPayment) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Место выдачи наличных
(CashPickupPoint). Метод GetPaymentDetails() должен быть переопределен
для отображения места выдачи наличных.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;

// ===== Интерфейсы (для множественного наследования) =====
public interface IPaymentLogger
{
    void LogPayment(decimal amount);
}

public interface IPaymentRefund
{
    void RefundPayment(decimal amount);
}

// Базовый класс — способ оплаты
public class PaymentMethod
{
    // Приватные поля — под свойства
    private int _paymentMethodId;
    private string _methodName;
    private decimal _minAmount;

    // === НОВЫЕ атрибуты базового класса ===
    private string _currency;
    private decimal _processingFee;
    private bool _isActive;

    // Свойство PaymentMethodId с геттером и сеттером
    public int PaymentMethodId
    {
        get { return _paymentMethodId; }
        set { _paymentMethodId = value; }
    }

    // Свойство MethodName с геттером и сеттером
    public string MethodName
    {
        get { return _methodName; }
        set { _methodName = value; }
    }

    // Свойство MinAmount с геттером, сеттером и валидацией
    public decimal MinAmount
    {
        get { return _minAmount; }
        set
        {
            if (value >= 0)
                _minAmount = value;
            else
                Console.WriteLine("Минимальная сумма не может быть отрицательной!");
        }
    }

    // === НОВЫЕ свойства ===
    public string Currency
    {
        get { return _currency; }
        set { _currency = value; }
    }

    public decimal ProcessingFee
    {
        get { return _processingFee; }
        set
        {
            if (value >= 0 && value <= 1)
                _processingFee = value;
            else
                Console.WriteLine("Комиссия должна быть в диапазоне от 0 до 1!");
        }
    }

    public bool IsActive
    {
        get { return _isActive; }
        set { _isActive = value; }
    }

    // Конструктор — присваивание через свойства
    public PaymentMethod(int paymentMethodId, string methodName, decimal minAmount)
    {
        PaymentMethodId = paymentMethodId;
        MethodName = methodName;
        MinAmount = minAmount;
        Currency = "RUB";
        ProcessingFee = 0m;
        IsActive = true;
    }

    // ===== Методы =====

    public virtual void ProcessPayment(decimal amount)
    {
        if (!IsActive)
        {
            Console.WriteLine($"[{MethodName}] Способ оплаты отключён.");
            return;
        }
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Базовый] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Базовый] Платёж на сумму {amount} успешно обработан через «{MethodName}».");
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"[Базовый] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}");
    }

    // === НОВЫЕ методы ===
    public void Activate()
    {
        IsActive = true;
        Console.WriteLine($"[{MethodName}] Способ активирован.");
    }

    public void Deactivate()
    {
        IsActive = false;
        Console.WriteLine($"[{MethodName}] Способ деактивирован.");
    }

    public string GetBaseInfo()
    {
        return $"ID: {PaymentMethodId}, Способ: {MethodName}, Валюта: {Currency}, Комиссия: {ProcessingFee:P0}, Активен: {(IsActive ? "да" : "нет")}";
    }

    // ===== Взаимодействие объектов =====

    public void CompareWith(PaymentMethod other)
    {
        Console.WriteLine($"Сравнение «{MethodName}» и «{other.MethodName}»:");
        if (MinAmount < other.MinAmount)
            Console.WriteLine($"  «{MethodName}» доступнее (мин. сумма ниже).");
        else if (MinAmount > other.MinAmount)
            Console.WriteLine($"  «{other.MethodName}» доступнее (мин. сумма ниже).");
        else
            Console.WriteLine("  Минимальные суммы равны.");
    }

    public virtual void ProcessPayment(decimal amount, PaymentMethod fallback)
    {
        if (CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[{MethodName}] Платёж {amount} обработан напрямую.");
        }
        else
        {
            Console.WriteLine($"[{MethodName}] Сумма {amount} меньше минимальной ({MinAmount}).");
            Console.WriteLine($"[{MethodName}] Передаю платёж способу «{fallback.MethodName}»...");
            fallback.ProcessPayment(amount);
        }
    }
}

// Производный класс
public class OnlinePayment : PaymentMethod, IPaymentLogger, IPaymentRefund
{
    private string _paymentUrl;

    // === НОВЫЕ атрибуты ===
    private string _provider;
    private bool _requires3DS;

    public string PaymentUrl
    {
        get { return _paymentUrl; }
        set { _paymentUrl = value; }
    }

    // === НОВЫЕ свойства ===
    public string Provider
    {
        get { return _provider; }
        set { _provider = value; }
    }

    public bool Requires3DS
    {
        get { return _requires3DS; }
        set { _requires3DS = value; }
    }

    public OnlinePayment(int paymentMethodId, string methodName, decimal minAmount, string paymentUrl)
        : base(paymentMethodId, methodName, minAmount)
    {
        PaymentUrl = paymentUrl;
        Provider = "Не указан";
        Requires3DS = true;
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!IsActive)
        {
            Console.WriteLine($"[Онлайн] Способ оплаты отключён.");
            return;
        }
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Онлайн] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Онлайн] Перенаправление на {PaymentUrl}...");
        Console.WriteLine($"[Онлайн] Провайдер: {Provider}, 3-D Secure: {(Requires3DS ? "включён" : "отключён")}");
        Console.WriteLine($"[Онлайн] Платёж на сумму {amount} успешно проведён.");
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Онлайн] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, URL: {PaymentUrl}");
    }

    // === НОВЫЕ методы ===
    public void StartSession()
    {
        Console.WriteLine($"[Онлайн] Сессия оплаты через «{Provider}» открыта.");
    }

    public void EndSession()
    {
        Console.WriteLine($"[Онлайн] Сессия оплаты через «{Provider}» закрыта.");
    }

    // === Реализация интерфейсов ===
    public void LogPayment(decimal amount)
    {
        Console.WriteLine($"[Лог] Онлайн-платёж на {amount} через «{Provider}» записан в журнал.");
    }

    public void RefundPayment(decimal amount)
    {
        Console.WriteLine($"[Возврат] Онлайн-платёж на {amount} возвращён через «{Provider}».");
    }
}

// Производный класс
public class BankTransfer : PaymentMethod, IPaymentLogger
{
    private string _bankData;
    private decimal _commissionRate;

    // === НОВЫЕ атрибуты ===
    private string _bankName;
    private int _transferDays;

    public string BankData
    {
        get { return _bankData; }
        set { _bankData = value; }
    }

    public decimal CommissionRate
    {
        get { return _commissionRate; }
        set
        {
            if (value >= 0 && value <= 1)
                _commissionRate = value;
            else
                Console.WriteLine("Комиссия должна быть в диапазоне от 0 до 1!");
        }
    }

    // === НОВЫЕ свойства ===
    public string BankName
    {
        get { return _bankName; }
        set { _bankName = value; }
    }

    public int TransferDays
    {
        get { return _transferDays; }
        set
        {
            if (value >= 0)
                _transferDays = value;
            else
                Console.WriteLine("Срок перевода не может быть отрицательным!");
        }
    }

    public BankTransfer(int paymentMethodId, string methodName, decimal minAmount, string bankData)
        : base(paymentMethodId, methodName, minAmount)
    {
        BankData = bankData;
        CommissionRate = 0.02m;
        BankName = "Не указан";
        TransferDays = 3;
    }

    public BankTransfer(int paymentMethodId, string methodName, decimal minAmount)
        : base(paymentMethodId, methodName, minAmount)
    {
        BankData = "ИНН не указан";
        CommissionRate = 0.02m;
        BankName = "Не указан";
        TransferDays = 3;
    }

    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * CommissionRate;
        Console.WriteLine($"[Перевод] Комиссия банка: {commission}. Итого к списанию: {amount + commission}");
        return amount >= MinAmount;
    }

    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"[Перевод] Банковские данные: {BankData}");
        base.ProcessPayment(amount);
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Перевод] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, Реквизиты: {BankData}, Комиссия: {CommissionRate:P0}");
    }

    // === НОВЫЕ методы ===
    public void EstimateTransferTime()
    {
        Console.WriteLine($"[Перевод] Ожидаемый срок зачисления через «{BankName}»: {TransferDays} дн.");
    }

    public void PrintBankDetails()
    {
        Console.WriteLine($"[Перевод] Банк: {BankName}, Реквизиты: {BankData}, Срок: {TransferDays} дн.");
    }

    // === Реализация интерфейса ===
    public void LogPayment(decimal amount)
    {
        Console.WriteLine($"[Лог] Банковский перевод на {amount} через «{BankName}» записан в журнал.");
    }
}

// Производный класс
public class CashPayment : PaymentMethod
{
    private string _cashPickupPoint;

    // === НОВЫЕ атрибуты ===
    private decimal _maxCashAmount;
    private bool _requiresID;

    public string CashPickupPoint
    {
        get { return _cashPickupPoint; }
        set { _cashPickupPoint = value; }
    }

    // === НОВЫЕ свойства ===
    public decimal MaxCashAmount
    {
        get { return _maxCashAmount; }
        set
        {
            if (value >= 0)
                _maxCashAmount = value;
            else
                Console.WriteLine("Максимальная сумма не может быть отрицательной!");
        }
    }

    public bool RequiresID
    {
        get { return _requiresID; }
        set { _requiresID = value; }
    }

    public CashPayment(int paymentMethodId, string methodName, decimal minAmount, string cashPickupPoint)
        : base(paymentMethodId, methodName, minAmount)
    {
        CashPickupPoint = cashPickupPoint;
        MaxCashAmount = 100000m;
        RequiresID = true;
    }

    public CashPayment(int paymentMethodId, string methodName, decimal minAmount)
        : base(paymentMethodId, methodName, minAmount)
    {
        CashPickupPoint = "Место выдачи не назначено";
        MaxCashAmount = 100000m;
        RequiresID = true;
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Наличные] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, Место выдачи: {CashPickupPoint}");
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Наличные] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Наличные] Получите наличные в пункте: {CashPickupPoint}");
        Console.WriteLine($"[Наличные] Платёж на сумму {amount} подтверждён.");
    }

    // === НОВЫЕ методы ===
    public bool CheckCashLimit(decimal amount)
    {
        if (amount > MaxCashAmount)
        {
            Console.WriteLine($"[Наличные] Сумма {amount} превышает лимит выдачи ({MaxCashAmount}).");
            return false;
        }
        return true;
    }

    public void PrintPickupInfo()
    {
        Console.WriteLine($"[Наличные] Пункт выдачи: {CashPickupPoint}, Лимит: {MaxCashAmount}, Требуется ID: {(RequiresID ? "да" : "нет")}");
    }
}

// ===== СЛОЖНОЕ НАСЛЕДОВАНИЕ =====
// 2-й уровень: наследник CashPayment
public class GiftCertificatePayment : CashPayment
{
    // === Новые атрибуты ===
    private string _certificateCode;
    private DateTime _expirationDate;

    public string CertificateCode
    {
        get { return _certificateCode; }
        set { _certificateCode = value; }
    }

    public DateTime ExpirationDate
    {
        get { return _expirationDate; }
        set { _expirationDate = value; }
    }

    public GiftCertificatePayment(int paymentMethodId, string methodName, decimal minAmount, string cashPickupPoint, string certificateCode, DateTime expirationDate)
        : base(paymentMethodId, methodName, minAmount, cashPickupPoint)
    {
        CertificateCode = certificateCode;
        ExpirationDate = expirationDate;
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"[Сертификат] Код: {CertificateCode}, Действителен до: {ExpirationDate:dd.MM.yyyy}");
    }

    // === Новые методы ===
    public bool CheckCertificate()
    {
        if (DateTime.Now > ExpirationDate)
        {
            Console.WriteLine($"[Сертификат] Код {CertificateCode} просрочен ({ExpirationDate:dd.MM.yyyy}).");
            return false;
        }
        Console.WriteLine($"[Сертификат] Код {CertificateCode} действителен.");
        return true;
    }

    public void PrintCertificateInfo()
    {
        Console.WriteLine($"[Сертификат] Код: {CertificateCode}, Действителен до: {ExpirationDate:dd.MM.yyyy}, Стоимость: {MinAmount}");
    }
}

// ===== СЛОЖНОЕ НАСЛЕДОВАНИЕ =====
// 3-й уровень: наследник GiftCertificatePayment
public class GiftCardPayment : GiftCertificatePayment, IPaymentLogger, IPaymentRefund
{
    // === Новые атрибуты ===
    private string _cardNumber;
    private decimal _balance;

    public string CardNumber
    {
        get { return _cardNumber; }
        set { _cardNumber = value; }
    }

    public decimal Balance
    {
        get { return _balance; }
        set
        {
            if (value >= 0)
                _balance = value;
            else
                Console.WriteLine("Баланс не может быть отрицательным!");
        }
    }

    public GiftCardPayment(int paymentMethodId, string methodName, decimal minAmount, string cashPickupPoint, string certificateCode, DateTime expirationDate, string cardNumber)
        : base(paymentMethodId, methodName, minAmount, cashPickupPoint, certificateCode, expirationDate)
    {
        CardNumber = cardNumber;
        Balance = 0m;
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"[Карта] Номер: {CardNumber}, Баланс: {Balance:C}");
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Карта] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        if (Balance < amount)
        {
            Console.WriteLine($"[Карта] Недостаточно средств на карте (баланс {Balance}, требуется {amount}).");
            return;
        }
        Balance -= amount;
        Console.WriteLine($"[Карта] Платёж на {amount} с карты {CardNumber} подтверждён. Новый баланс: {Balance:C}");
    }

    // === Новые методы ===
    public void TopUpCard(decimal amount)
    {
        if (amount > 0)
        {
            Balance += amount;
            Console.WriteLine($"[Карта] Карта {CardNumber} пополнена на {amount}. Баланс: {Balance:C}");
        }
        else
        {
            Console.WriteLine("[Карта] Сумма пополнения должна быть положительной.");
        }
    }

    public void PrintCardInfo()
    {
        Console.WriteLine($"[Карта] Номер: {CardNumber}, Баланс: {Balance:C}, Код сертификата: {CertificateCode}");
    }

    // === Реализация интерфейсов ===
    public void LogPayment(decimal amount)
    {
        Console.WriteLine($"[Лог] Оплата картой {CardNumber} на {amount} записана в журнал.");
    }

    public void RefundPayment(decimal amount)
    {
        Balance += amount;
        Console.WriteLine($"[Возврат] На карту {CardNumber} возвращено {amount}. Баланс: {Balance:C}");
    }
}

// ==== top-level код (точка входа) ====
PaymentMethod[] methods = new PaymentMethod[]
{
    new OnlinePayment(1, "Карта онлайн", 100m, "https://pay.example.com"),
    new BankTransfer(2, "Банковский перевод", 500m, "ИНН: 7707083893"),
    new CashPayment(3, "Наличные в кассе", 50m, "Касса №5, ул. Ленина 10")
};

Console.WriteLine("══════════ Детали всех способов оплаты ══════════");
foreach (var method in methods)
{
    method.GetPaymentDetails();
}

Console.WriteLine("\n══════════ Взаимодействие 1: сравнение объектов ══════════");
methods[0].CompareWith(methods[1]);
methods[1].CompareWith(methods[2]);

Console.WriteLine("\n══════════ Взаимодействие 2: делегирование платежа ══════════");
methods[0].ProcessPayment(30m, methods[2]);
Console.WriteLine();
methods[1].ProcessPayment(200m, methods[0]);

Console.WriteLine("\n══════════ Проверка второго конструктора ══════════");
BankTransfer lateBank = new BankTransfer(4, "Перевод позже", 300m);
lateBank.BankData = "ИНН: 7736050003";
lateBank.GetPaymentDetails();

// ===== НОВЫЙ БЛОК: Сложное наследование =====
Console.WriteLine("\n══════════ Сложное наследование: Cash → Certificate → Card ══════════");

GiftCardPayment giftCard = new GiftCardPayment(
    5,
    "Подарочная карта",
    500m,
    "Касса №1, ул. Ленина 10",
    "CERT-2024-001",
    new DateTime(2026, 12, 31),
    "4276-XXXX-XXXX-1234"
);

// Цепочка вызовов — методы всех трёх уровней
giftCard.PrintPickupInfo();        // из CashPayment
giftCard.PrintCertificateInfo();   // из GiftCertificatePayment
giftCard.PrintCardInfo();          // из GiftCardPayment

giftCard.GetPaymentDetails();      // накапливает вывод из трёх уровней

// Работа с картой
giftCard.TopUpCard(1500m);
giftCard.ProcessPayment(700m);

// ===== НОВЫЙ БЛОК: Множественное наследование =====
Console.WriteLine("\n══════════ Множественное наследование: класс + интерфейсы ══════════");

OnlinePayment online = (OnlinePayment)methods[0];
online.Provider = "Сбербанк-Онлайн";
online.StartSession();
online.LogPayment(1000m);         // из IPaymentLogger
online.RefundPayment(500m);       // из IPaymentRefund
online.EndSession();

Console.WriteLine();
giftCard.LogPayment(700m);        // из IPaymentLogger
giftCard.RefundPayment(200m);     // из IPaymentRefund

// ===== НОВЫЙ БЛОК: Новые атрибуты базового класса =====
Console.WriteLine("\n══════════ Новые атрибуты базового класса ══════════");
methods[0].Deactivate();
methods[0].ProcessPayment(200m);
methods[0].Activate();
methods[0].ProcessPayment(200m);
Console.WriteLine(methods[0].GetBaseInfo());
```
