Описание задачи:
Создать базовый класс PaymentMethod в C#, который будет представлять
различные способы оплаты. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу PaymentMethod:
• Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты
(MethodName), Минимальная сумма (MinAmount).
• Методы:
o ProcessPayment(decimal amount): метод для обработки платежа
указанной суммы.
o CheckMinimumAmount(decimal amount): метод для проверки
минимальной суммы платежа.
o GetPaymentDetails(): метод для получения деталей способа оплаты.
Требования к производным классам:
1. ОнлайнОплата (OnlinePayment): Должен содержать дополнительные
атрибуты, такие как URL платежной системы (PaymentUrl).
Метод ProcessPayment() должен быть переопределен для включения URL
платежной системы в процесс оплаты.
2. БанковскийПеревод (BankTransfer): Должен содержать дополнительные
атрибуты, такие как Банковские данные (BankData).
Метод CheckMinimumAmount() должен быть переопределен для проверки
минимальной суммы платежа с учетом банковских комиссий.
3. Наличные (CashPayment) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Место выдачи наличных
(CashPickupPoint). Метод GetPaymentDetails() должен быть переопределен
для отображения места выдачи наличных.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый класс — способ оплаты
public class PaymentMethod
{
    // Приватные поля — под свойства
    private int _paymentMethodId;
    private string _methodName;
    private decimal _minAmount;

    // Свойство PaymentMethodId с геттером и сеттером
    public int PaymentMethodId
    {
        get { return _paymentMethodId; }
        set { _paymentMethodId = value; }
    }

    // Свойство MethodName с геттером и сеттером
    public string MethodName
    {
        get { return _methodName; }
        set { _methodName = value; }
    }

    // Свойство MinAmount с геттером, сеттером и валидацией
    public decimal MinAmount
    {
        get { return _minAmount; }
        set
        {
            if (value >= 0)
                _minAmount = value;
            else
                Console.WriteLine("Минимальная сумма не может быть отрицательной!");
        }
    }

    // Конструктор — присваивание через свойства
    public PaymentMethod(int paymentMethodId, string methodName, decimal minAmount)
    {
        PaymentMethodId = paymentMethodId;
        MethodName = methodName;
        MinAmount = minAmount;
    }

    // ===== Методы =====

    public virtual void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Базовый] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Базовый] Платёж на сумму {amount} успешно обработан через «{MethodName}».");
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"[Базовый] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}");
    }

    // ===== Взаимодействие объектов =====

    // 1. Сравнение с другим способом оплаты (объект читает свойства другого)
    public void CompareWith(PaymentMethod other)
    {
        Console.WriteLine($"Сравнение «{MethodName}» и «{other.MethodName}»:");
        if (MinAmount < other.MinAmount)
            Console.WriteLine($"  «{MethodName}» доступнее (мин. сумма ниже).");
        else if (MinAmount > other.MinAmount)
            Console.WriteLine($"  «{other.MethodName}» доступнее (мин. сумма ниже).");
        else
            Console.WriteLine("  Минимальные суммы равны.");
    }

    // 2. Делегирование платежа другому способу (передача работы)
    public virtual void ProcessPayment(decimal amount, PaymentMethod fallback)
    {
        if (CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[{MethodName}] Платёж {amount} обработан напрямую.");
        }
        else
        {
            Console.WriteLine($"[{MethodName}] Сумма {amount} меньше минимальной ({MinAmount}).");
            Console.WriteLine($"[{MethodName}] Передаю платёж способу «{fallback.MethodName}»...");
            fallback.ProcessPayment(amount);
        }
    }
}

// Производный класс
public class OnlinePayment : PaymentMethod
{
    private string _paymentUrl;

    // Свойство PaymentUrl с геттером и сеттером
    public string PaymentUrl
    {
        get { return _paymentUrl; }
        set { _paymentUrl = value; }
    }

    public OnlinePayment(int paymentMethodId, string methodName, decimal minAmount, string paymentUrl)
        : base(paymentMethodId, methodName, minAmount)
    {
        PaymentUrl = paymentUrl;
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Онлайн] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Онлайн] Перенаправление на {PaymentUrl}...");
        Console.WriteLine($"[Онлайн] Платёж на сумму {amount} успешно проведён.");
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Онлайн] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, URL: {PaymentUrl}");
    }
}

// Производный класс
public class BankTransfer : PaymentMethod
{
    private string _bankData;
    private decimal _commissionRate;

    // Свойство BankData с геттером и сеттером
    public string BankData
    {
        get { return _bankData; }
        set { _bankData = value; }
    }

    // Свойство CommissionRate с геттером и сеттером
    public decimal CommissionRate
    {
        get { return _commissionRate; }
        set
        {
            if (value >= 0 && value <= 1)
                _commissionRate = value;
            else
                Console.WriteLine("Комиссия должна быть в диапазоне от 0 до 1!");
        }
    }

    // Конструктор 1 — с реквизитами
    public BankTransfer(int paymentMethodId, string methodName, decimal minAmount, string bankData)
        : base(paymentMethodId, methodName, minAmount)
    {
        BankData = bankData;
        CommissionRate = 0.02m;
    }

    // Конструктор 2 — без реквизитов (устанавливаются позже через свойство)
    public BankTransfer(int paymentMethodId, string methodName, decimal minAmount)
        : base(paymentMethodId, methodName, minAmount)
    {
        BankData = "ИНН не указан";
        CommissionRate = 0.02m;
    }

    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * CommissionRate;
        Console.WriteLine($"[Перевод] Комиссия банка: {commission}. Итого к списанию: {amount + commission}");
        return amount >= MinAmount;
    }

    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"[Перевод] Банковские данные: {BankData}");
        base.ProcessPayment(amount);
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Перевод] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, Реквизиты: {BankData}, Комиссия: {CommissionRate:P0}");
    }
}

// Производный класс
public class CashPayment : PaymentMethod
{
    private string _cashPickupPoint;

    // Свойство CashPickupPoint с геттером и сеттером
    public string CashPickupPoint
    {
        get { return _cashPickupPoint; }
        set { _cashPickupPoint = value; }
    }

    // Конструктор 1 — с местом выдачи
    public CashPayment(int paymentMethodId, string methodName, decimal minAmount, string cashPickupPoint)
        : base(paymentMethodId, methodName, minAmount)
    {
        CashPickupPoint = cashPickupPoint;
    }

    // Конструктор 2 — без места выдачи (назначается позже)
    public CashPayment(int paymentMethodId, string methodName, decimal minAmount)
        : base(paymentMethodId, methodName, minAmount)
    {
        CashPickupPoint = "Место выдачи не назначено";
    }

    public override void GetPaymentDetails()
    {
        Console.WriteLine($"[Наличные] ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount}, Место выдачи: {CashPickupPoint}");
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[Наличные] Сумма {amount} меньше минимальной ({MinAmount}). Платёж отклонён.");
            return;
        }
        Console.WriteLine($"[Наличные] Получите наличные в пункте: {CashPickupPoint}");
        Console.WriteLine($"[Наличные] Платёж на сумму {amount} подтверждён.");
    }
}

// ==== top-level код (точка входа) ====
PaymentMethod[] methods = new PaymentMethod[]
{
    new OnlinePayment(1, "Карта онлайн", 100m, "https://pay.example.com"),
    new BankTransfer(2, "Банковский перевод", 500m, "ИНН: 7707083893"),
    new CashPayment(3, "Наличные в кассе", 50m, "Касса №5, ул. Ленина 10")
};

Console.WriteLine("══════════ Детали всех способов оплаты ══════════");
foreach (var method in methods)
{
    method.GetPaymentDetails();
}

Console.WriteLine("\n══════════ Взаимодействие 1: сравнение объектов ══════════");
methods[0].CompareWith(methods[1]);
methods[1].CompareWith(methods[2]);

Console.WriteLine("\n══════════ Взаимодействие 2: делегирование платежа ══════════");
// Платёж 30 меньше минимума онлайн-оплаты (100) — передаём наличным (мин. 50)
methods[0].ProcessPayment(30m, methods[2]);

Console.WriteLine();
// Платёж 200 меньше минимума банковского перевода (500) — передаём онлайн-оплате (мин. 100)
methods[1].ProcessPayment(200m, methods[0]);

Console.WriteLine("\n══════════ Проверка второго конструктора ══════════");
BankTransfer lateBank = new BankTransfer(4, "Перевод позже", 300m);
lateBank.BankData = "ИНН: 7736050003";
lateBank.GetPaymentDetails();
```
